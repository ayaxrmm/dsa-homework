````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База

#1
cities_clean = []
for c in cities:
  cities_clean.append(c.strip().title())
print(cities_clean)

#2
amounts_int = []
for a in amounts:
  amounts_int.append(int(a))
print(amounts_int)

#3
print(len(amounts))
print(sum(amounts_int))
print(round(sum(amounts_int) / len(amounts)))

#4
c2, c3 = 0, 0
for i in range(len(paid)):
  if paid[i] == False:
    c2 +=1
    c3 += amounts_int[i]
print(f"{c2} заказов не оплачено на сумму - {c3}\n")

#5
for c, a, p in zip(cities, amounts, paid):
    status = "оплачен" if p else "НЕ оплачен"
    print(f"{c:<10} {a:>7} ₸  {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
#6
almaty_sum, astana_sum = 0, 0
for i in range(len(amounts)):
  if paid[i] == True and cities_clean[i] == "Almaty":
    almaty_sum += amounts_int[i]
  elif paid[i] == True and cities_clean[i] == "Astana":
    astana_sum += amounts_int[i]
print(f"выручка по Алматы: {almaty_sum}, выручка по Астане: {astana_sum}")

#7
m1 = 0
for a, c in zip(amounts_int, cities_clean):
  m1 += 1
  if a == max(amounts_int):
    print(m1, c)

#8
s1 = 0
for i in range(len(days)):
  if days[i] == "сб" or days[i] == "вс":
    s1 += 1
print(f"{round(s1/len(days)*100)} %")

#9
goal, a, month, total = 400000, 15000, 0, 0
while total < goal:
  total += a
  month += 1
print(f"через {month} месяцев накопит")
month, total = 0, 0
while total < goal:
  total += a
  a += 1000
  month += 1
print(f"через {month} месяцев накопит если прибавлять 1000 каждый месяц")
````
